```python
import pandas as pd

df = pd.read_csv("Untitled form.csv")

print("Total respondents:", len(df))
```

```python
print("Duplicate responses:", df.duplicated().sum())
```

```python
consent_col = "Do you agree to participate in the survey?"

print(df[consent_col].value_counts(dropna=False))
```

```python
df[consent_col] = df[consent_col].replace("Yess", "Yes")

print(df[consent_col].value_counts(dropna=False))
```

```python
age_col = "Q1. What is your age group?"

print(df[age_col].value_counts(dropna=False))
```

```python
df["Eligible"] = (
    df[consent_col].eq("Yes") &
    df[age_col].ne("Below 18") &
    df[age_col].notna()
)

print(df["Eligible"].value_counts())
```

```python
print(df.isnull().sum())
```

```python
print("TOTAL RESPONDENTS:", len(df))
print("DUPLICATES:", df.duplicated().sum())
print("ELIGIBLE:", df["Eligible"].sum())
print("NOT ELIGIBLE:", (~df["Eligible"]).sum())
print("MISSING CONSENT:", df[consent_col].isnull().sum())
```
