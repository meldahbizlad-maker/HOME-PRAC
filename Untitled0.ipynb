```python
df_clean = df[df["Eligible"]].copy()

print("Clean dataset:", df_clean.shape)
```

```python
for col in df_clean.columns:
    if any(word in col.lower() for word in ["rent", "accommodation", "food", "transport"]):
        print(col)
```

```python
for col in df_clean.columns:
    if col.startswith(("Q10.", "Q11.", "Q12.")):
        print("\n", col)
        print(df_clean[col].value_counts(dropna=False))
```

```python
food_col = [c for c in df_clean.columns if c.startswith("Q10.")][0]
rent_col = [c for c in df_clean.columns if c.startswith("Q11.")][0]
transport_col = [c for c in df_clean.columns if c.startswith("Q12.")][0]

df_clean["Food_USD"] = df_clean[food_col].map({
    "Less than $50": 25,
    "$50-$100": 75,
    "$100-$300": 200,
    "Above $300": 350
})

df_clean["Rent_USD"] = df_clean[rent_col].map({
    "No rent": 0,
    "Less than $100": 50,
    "$100-$250": 175,
    "Above $250": 300
})

df_clean["Transport_USD"] = df_clean[transport_col].map({
    "Less than $20": 10,
    "$20-$50": 35,
    "Above $50": 60
})

print(df_clean[["Food_USD", "Rent_USD", "Transport_USD"]].head())
```

```python
df_clean["Total_Essential_Expenditure_USD"] = (
    df_clean["Food_USD"] +
    df_clean["Rent_USD"] +
    df_clean["Transport_USD"]
)

print(df_clean[[
    "Food_USD",
    "Rent_USD",
    "Transport_USD",
    "Total_Essential_Expenditure_USD"
]].head())
```

```python
df_clean["Food_Share_Percent"] = (
    df_clean["Food_USD"] /
    df_clean["Total_Essential_Expenditure_USD"]
) * 100

print(df_clean[["Food_USD", "Total_Essential_Expenditure_USD", "Food_Share_Percent"]].head())
```

```python
df_clean["Food_Exceeds_50pct"] = df_clean["Food_Share_Percent"] > 50

print(df_clean["Food_Exceeds_50pct"].value_counts())
```

```python
mean_expenditure = df_clean["Total_Essential_Expenditure_USD"].mean()

print("Mean monthly essential expenditure: $", round(mean_expenditure, 2))
```

```python
mean_rent = df_clean.loc[df_clean["Rent_USD"] > 0, "Rent_USD"].mean()

print("Mean monthly rent: $", round(mean_rent, 2))
```

```python
#weighting demonstration
import pandas as pd

weight_demo = pd.DataFrame({
    "Area_Type": ["Urban"]*5 + ["Rural"]*5,
    "Food_Exceeds_50pct": [True, True, True, False, False,
                           True, True, True, True, False]
})

weight_demo["Selection_Probability"] = weight_demo["Area_Type"].map({
    "Urban": 0.10,
    "Rural": 0.20
})

weight_demo["Base_Weight"] = 1 / weight_demo["Selection_Probability"]

print(weight_demo)
```

```python
unweighted = weight_demo["Food_Exceeds_50pct"].mean() * 100

weighted = (
    weight_demo["Food_Exceeds_50pct"] * weight_demo["Base_Weight"]
).sum() / weight_demo["Base_Weight"].sum() * 100

print("Unweighted percentage:", round(unweighted, 2), "%")
print("Weighted percentage:", round(weighted, 2), "%")
```

```python
# Create rent eligibility flag
df_clean["Rent_Payer_Eligible"] = df_clean["Rent_USD"] > 0

# Create benchmark summary
results = []

for area, data in [("Overall", df_clean)] + list(df_clean.groupby(domain_col)):

    results.append({
        "Area_Type": area,
        "Mean_Essential_Expenditure_USD":
            data["Total_Essential_Expenditure_USD"].mean(),
        "Food_Burden_Percent":
            data["Food_Exceeds_50pct"].mean() * 100,
        "Mean_Rent_Among_Rent_Payers_USD":
            data.loc[data["Rent_USD"] > 0, "Rent_USD"].mean(),
        "Number_of_Households": len(data),
        "Rent_Paying_Households":
            data["Rent_Payer_Eligible"].sum()
    })

python_benchmark = pd.DataFrame(results).round(2)

# Weighting summary
weighting_summary = pd.DataFrame({
    "Measure": ["Unweighted percentage", "Weighted percentage"],
    "Percentage": [unweighted, weighted]
})

# Export all four files
df_clean.to_csv("clean_pilot.csv", index=False)
python_benchmark.to_csv("python_benchmark.csv", index=False)
weight_demo.to_csv("weighting_demo_case_level.csv", index=False)
weighting_summary.to_csv("weighting_demo_summary.csv", index=False)

print("All four CSV files saved successfully!")
print(python_benchmark)
print(weighting_summary)
```
